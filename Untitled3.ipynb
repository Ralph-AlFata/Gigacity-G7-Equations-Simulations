```python
import os
import sys
import matplotlib.pyplot as plt

# 1. Configure Hugging Face
os.environ["HF_HOME"] = "/tmp/hf_auth"
os.environ["HF_HUB_CACHE"] = "/workspace/alpamayo1.5/.hf_cache/hub"
os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "0"

# 2. Verify Python environment
print("Python:", sys.executable)

# 3. Import NVIDIA's dataset loader
from alpamayo1_5.load_physical_aiavdataset import (
    load_physical_aiavdataset
)

# 4. Load the exact scenario used in test_inference.py
clip_id = "030c760c-ae38-49aa-9ad8-f5650a545d26"
t0_us = 5_100_000

print("Loading driving scenario...")
data = load_physical_aiavdataset(clip_id, t0_us=t0_us)

# Shape: (4 cameras, 4 frames, 3, H, W)
frames = data["image_frames"]
camera_indices = data["camera_indices"]

print("Image tensor shape:", frames.shape)
print("Camera indices:", camera_indices.tolist())

# 5. Map camera IDs to names
camera_names = {
    0: "Front Left (120°)",
    1: "Front Wide (120°)",
    2: "Front Right (120°)",
    6: "Front Telephoto (30°)",
}

# 6. Visualize all 16 input images
num_cameras, num_frames = frames.shape[:2]

fig, axes = plt.subplots(
    num_cameras,
    num_frames,
    figsize=(20, 12)
)

for i in range(num_cameras):
    cam_idx = int(camera_indices[i])
    cam_name = camera_names.get(cam_idx, f"Camera {cam_idx}")

    for j in range(num_frames):
        # Convert (C, H, W) tensor to (H, W, C)
        image = frames[i, j].permute(1, 2, 0).cpu().numpy()

        # Actual frame timestamp relative to prediction time
        timestamp = (
            int(data["absolute_timestamps"][i, j]) - t0_us
        ) / 1_000_000

        axes[i, j].imshow(image)
        axes[i, j].set_title(
            f"{cam_name}\nt = {timestamp:+.2f}s",
            fontsize=10
        )
        axes[i, j].axis("off")

fig.suptitle(
    "Alpamayo 1.5 — Input Camera Frames",
    fontsize=18
)

plt.tight_layout()

# 7. Save visualization to persistent storage
output_path = "/workspace/alpamayo1.5/input_images.png"

plt.savefig(
    output_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(f"Saved visualization to: {output_path}")
```

```python

import os
import numpy as np
import torch
import matplotlib.pyplot as plt

from alpamayo1_5 import helper
from alpamayo1_5.models.alpamayo1_5 import Alpamayo1_5
from alpamayo1_5.load_physical_aiavdataset import (
    load_physical_aiavdataset
)

# Hugging Face configuration
os.environ["HF_HOME"] = "/tmp/hf_auth"
os.environ["HF_HUB_CACHE"] = "/workspace/alpamayo1.5/.hf_cache/hub"
os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "0"

# 1. Load the same driving scenario
clip_id = "030c760c-ae38-49aa-9ad8-f5650a545d26"
data = load_physical_aiavdataset(clip_id, t0_us=5_100_000)

# 2. Prepare the model inputs
messages = helper.create_message(
    frames=data["image_frames"].flatten(0, 1),
    camera_indices=data["camera_indices"]
)

print("Loading Alpamayo 1.5...")
model = Alpamayo1_5.from_pretrained(
    "nvidia/Alpamayo-1.5-10B",
    dtype=torch.bfloat16
).to("cuda")
model.eval()

processor = helper.get_processor(model.tokenizer)

inputs = processor.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=False,
    continue_final_message=True,
    return_dict=True,
    return_tensors="pt",
)

model_inputs = {
    "tokenized_data": inputs,
    "ego_history_xyz": data["ego_history_xyz"],
    "ego_history_rot": data["ego_history_rot"],
}

model_inputs = helper.to_device(model_inputs, "cuda")

# 3. Generate the predicted trajectory
torch.cuda.manual_seed_all(42)

with torch.no_grad(), torch.autocast("cuda", dtype=torch.bfloat16):
    pred_xyz, pred_rot, extra = (
        model.sample_trajectories_from_data_with_vlm_rollout(
            data=model_inputs,
            top_p=0.98,
            temperature=0.6,
            num_traj_samples=1,
            max_generation_length=256,
            return_extra=True,
        )
    )

# 4. Extract XY positions
# Each trajectory is represented in the vehicle's
# local coordinate system at prediction time.

history_xy = data["ego_history_xyz"][0, 0, :, :2].cpu().numpy()
gt_xy = data["ego_future_xyz"][0, 0, :, :2].cpu().numpy()
pred_xy = pred_xyz[0, 0, 0, :, :2].detach().cpu().numpy()

# 5. Calculate average displacement error
errors = np.linalg.norm(pred_xy - gt_xy, axis=1)
ade = errors.mean()

print("Chain-of-Causation:")
print(extra["cot"][0])
print(f"Average displacement error: {ade:.3f} m")

# 6. Plot trajectories
fig, ax = plt.subplots(figsize=(10, 8))

ax.plot(
    history_xy[:, 0], history_xy[:, 1],
    "o-", label="Past vehicle trajectory"
)

ax.plot(
    gt_xy[:, 0], gt_xy[:, 1],
    "-", linewidth=2.5, label="Ground truth (6.4 s)"
)

ax.plot(
    pred_xy[:, 0], pred_xy[:, 1],
    "--", linewidth=2.5, label="Alpamayo prediction (6.4 s)"
)

# Current vehicle position
ax.scatter(
    0, 0, s=150, marker="*",
    label="Current vehicle position", zorder=5
)

# Future endpoints
ax.scatter(gt_xy[-1, 0], gt_xy[-1, 1], marker="x", s=90)
ax.scatter(pred_xy[-1, 0], pred_xy[-1, 1], marker="x", s=90)

ax.set_xlabel("Local X (meters)")
ax.set_ylabel("Local Y (meters)")
ax.set_title(
    f"Alpamayo 1.5: Predicted vs Actual Trajectory\n"
    f"ADE = {ade:.3f} meters"
)

ax.axis("equal")
ax.grid(True, alpha=0.3)
ax.legend()

plt.tight_layout()

# 7. Save results to persistent storage
output_dir = "/workspace/alpamayo1.5"

plt.savefig(
    f"{output_dir}/trajectory_comparison.png",
    dpi=150,
    bbox_inches="tight"
)

np.savez_compressed(
    f"{output_dir}/trajectory_results.npz",
    history_xy=history_xy,
    ground_truth_xy=gt_xy,
    predicted_xy=pred_xy,
    displacement_errors=errors,
)

plt.show()
```

```python
from alpamayo1_5 import helper
from alpamayo1_5.models.alpamayo1_5 import Alpamayo1_5
from alpamayo1_5.load_physical_aiavdataset import (
    load_physical_aiavdataset
)

# Reuse the model if already loaded in this notebook.
if "model" not in globals() or "model_inputs" not in globals():
    clip_id = "030c760c-ae38-49aa-9ad8-f5650a545d26"

    data = load_physical_aiavdataset(
        clip_id, t0_us=5_100_000
    )

    messages = helper.create_message(
        frames=data["image_frames"].flatten(0, 1),
        camera_indices=data["camera_indices"]
    )

    model = Alpamayo1_5.from_pretrained(
        "nvidia/Alpamayo-1.5-10B",
        dtype=torch.bfloat16
    ).to("cuda")

    processor = helper.get_processor(model.tokenizer)

    inputs = processor.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=False,
        continue_final_message=True,
        return_dict=True,
        return_tensors="pt",
    )

    model_inputs = helper.to_device({
        "tokenized_data": inputs,
        "ego_history_xyz": data["ego_history_xyz"],
        "ego_history_rot": data["ego_history_rot"],
    }, "cuda")

model.eval()

def run_inference():
    with torch.inference_mode(), torch.autocast(
        "cuda", dtype=torch.bfloat16
    ):
        return model.sample_trajectories_from_data_with_vlm_rollout(
            data=model_inputs,
            top_p=0.98,
            temperature=0.6,
            num_traj_samples=1,
            max_generation_length=256,
            return_extra=True,
        )

# Warmup
for _ in range(2):
    run_inference()
    torch.cuda.synchronize()

torch.cuda.reset_peak_memory_stats()

# Measure individual requests
N_RUNS = 20
latencies = []

for i in range(N_RUNS):
    torch.cuda.synchronize()
    start = time.perf_counter()

    result = run_inference()

    torch.cuda.synchronize()
    elapsed = time.perf_counter() - start

    latencies.append(elapsed)
    del result

    print(f"Request {i+1:02d}: {elapsed:.3f} seconds")

latencies = np.array(latencies)

# Save measurements
pd.DataFrame({
    "request": range(1, N_RUNS + 1),
    "latency_s": latencies
}).to_csv(PROFILE_PATH, index=False)

print("\n--- Inference Performance ---")
print(f"Mean latency: {latencies.mean():.3f} s")
print(f"P50 latency: {np.percentile(latencies, 50):.3f} s")
print(f"P90 latency: {np.percentile(latencies, 90):.3f} s")
print(f"P95 latency: {np.percentile(latencies, 95):.3f} s")
print(f"Throughput: {N_RUNS/latencies.sum():.3f} req/s")
print(
    f"Peak GPU memory: "
    f"{torch.cuda.max_memory_allocated()/1e9:.2f} GB"
)

plt.figure(figsize=(9, 4))
plt.hist(latencies, bins=10, edgecolor="white")
plt.axvline(
    np.median(latencies),
    linestyle="--",
    label="Median latency"
)
plt.xlabel("Inference latency (seconds)")
plt.ylabel("Frequency")
plt.title("Alpamayo 1.5 — Warm Inference Latency")
plt.legend()
plt.show()
```

```python
latencies = pd.read_csv(PROFILE_PATH)["latency_s"].to_numpy()
```

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Request:
    id: int
    vehicle: int
    arrival: float
    deadline: float
    actual: float
    estimated: float
    downlink: float
    weight: float


def generate_requests(
    n_vehicles=4,
    seed=42,
    horizon_multiplier=45,
    period_multiplier=5,
    burst_probability=0.0,
    heterogeneous=False,
):
    rng = np.random.default_rng(seed)

    median = float(np.median(latencies))
    p90 = float(np.percentile(latencies, 90))

    # All times are in seconds
    horizon = horizon_multiplier * median
    period = period_multiplier * median

    requests = []

    for vehicle in range(n_vehicles):
        # Stagger requests from different vehicles
        t = vehicle * period / n_vehicles

        while t < horizon:
            factor = (
                float(rng.choice([0.7, 1.0, 1.5]))
                if heterogeneous else 1.0
            )

            actual = float(rng.choice(latencies)) * factor
            estimated = p90 * factor

            # Synthetic communication delays
            upload = 0.02 * median
            downlink = 0.02 * median

            # Deadlines relative to request creation
            budget = float(
                rng.choice([1.5, 2.5, 4.0])
            ) * median

            requests.append(Request(
                id=len(requests),
                vehicle=vehicle,
                arrival=t + upload,
                deadline=t + budget,
                actual=actual,
                estimated=estimated,
                downlink=downlink,
                weight=float(rng.choice([1, 2, 3])),
            ))

            step = max(
                0.1 * median,
                float(rng.normal(period, 0.1 * period))
            )

            if rng.random() < burst_probability:
                step *= 0.15

            t += step

    return sorted(
        requests, key=lambda r: (r.arrival, r.id)
    )


requests = generate_requests(n_vehicles=4)

print(f"Generated {len(requests)} requests")
print(f"Vehicles: {len(set(r.vehicle for r in requests))}")

display(
    pd.DataFrame([vars(r) for r in requests[:10]]).round(3)
)
```
